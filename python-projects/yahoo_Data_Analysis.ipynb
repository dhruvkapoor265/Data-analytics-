```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import yfinance as yf
```

```python
nsei = yf.download('^NSEI',
start = '2019-01-01',
end = '2025-07-24',
progress = False)
```

```python
nsei['Close'].plot(figsize=(14,10))
```

```python
nsei[['Close','Volume']].plot(subplots=True,figsize=(12,6))
```

```python
nsei.describe()
```

```python
nsei['Close_simple_rtn'] = nsei.Close.pct_change()
```

```python
nsei
```

```python
nsei[['Close','Close_simple_rtn']].plot(subplots=True,figsize=(12,6))
```

```python
nsei['Close_log_rtn'] = np.log(nsei.Close/nsei.Close.shift(1))
```

```python
nsei
```

```python
nsei[['Close_simple_rtn','Close_log_rtn']].plot(subplots=True,figsize=(12,6))
```

```python
nsei['Close_std']=nsei.Close.rolling(window=12).std()
```

```python
nsei
```

```python
corr_close = nsei.Close_simple_rtn.corr(nsei.Close_log_rtn)
corr_close
```

```python
nsei['volulog_rtn'] = np.log(nsei.Volume/nsei.Volume.shift(1))
```

```python
nsei['voluSim_rtn'] = nsei.Volume.pct_change()
```

```python
nsei
```

```python
nsei[['volulog_rtn','voluSim_rtn']].plot(subplots=True,figsize=(12,6))
```

```python
WINDOW_SIZE = 12
nsei['rolling_mean']=nsei.Close.rolling(window=WINDOW_SIZE).mean()
nsei['rolling_std']=nsei.Close.rolling(window=WINDOW_SIZE).std()
```

```python
nsei.to_excel('nsei.xlsx')
```

```python
corr_close_mstd = nsei.Close_simple_rtn.corr(nsei.rolling_std)
corr_close_mstd
```

```python
nsei[['rolling_mean','rolling_std']].plot(subplots=True,figsize=(12,6))
```

```python
WINDOW_SIZE = 12
nsei['Vrolling_mean']=nsei.Volume.rolling(window=WINDOW_SIZE).mean()
nsei['Vrolling_std']=nsei.Volume.rolling(window=WINDOW_SIZE).std()
```

```python
nsei[['Vrolling_mean','Vrolling_std']].plot(subplots=True,figsize=(12,6))
```

```python
#correlation of Volume
vcorr=nsei.Vrolling_mean.corr(nsei.Vrolling_std)
vcorr
```
