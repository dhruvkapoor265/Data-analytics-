```python
#import the numpy library
import numpy as np
#define the database
x=np.array([78,89,97,69,59,79,68,57])
y=np.array([125,137,156,112,107,136,123,108])
```

```python
x
```

```python
y
```

```python
def pearson_corrrelation(x,y):
  if len(x)==len(y):
    sum_xy=sum((x-x.mean())*(y-y.mean()))
    sum_x_squared=sum((x-x.mean())**2)
    sum_y_squared=sum((y-y.mean())**2)
    corr=sum_xy/np.sqrt(sum_x_squared*sum_y_squared)
  return corr
print(pearson_corrrelation(x,y))
print(pearson_corrrelation(x,x))
```

```python
print(np.corrcoef(x,y))
```

```python
import pandas as pd
from sklearn.datasets import load_diabetes
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#load the dataset with frame
df=load_diabetes(as_frame=True)
#conver into pandas dataframe
df=df.frame
#print first 5 rows
df.head()
```

```python
df.tail()
```

```python
df[20:50]
```

```python
#find the pearson correlation matrix
corr=df.corr(method='pearson')
corr
```

```python
#correlation between age and sex columns
c=np.corrcoef(df['age'],df['sex'])
print('correlation between age and sex\n',c)
```

```python
plt.figure(figsize=(10,8),dpi=500)
sns.heatmap(corr,annot=True,fmt=".3f",linewidth=.5)
plt.show()
```
